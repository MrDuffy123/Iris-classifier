# Exploratory Data Analysis

This notebook is used for exploratory data analysis (EDA) on the Iris dataset. We will visualize the dataset and understand the relationships between features.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
data = pd.read_csv('../data/raw/iris.csv')  # Adjust the path as necessary

# Display the first few rows of the dataset
data.head()
```

```python
# Visualize the distribution of each feature
sns.pairplot(data, hue='species')
plt.show()
```

```python
# Correlation matrix
correlation = data.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

## Conclusion

In this notebook, we performed exploratory data analysis on the Iris dataset. We visualized the relationships between features and examined the correlation matrix.
